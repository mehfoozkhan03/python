```python
def products(a,b,c,d):
    if d == "sum":
        return a+b+c
    
    else:
        val= a+b+c
        return val/3 
        
```

```python
ans = products(3,6.99,10.89,"sum")
ans2 = products(3,6.99,10.89,"avg")
print(ans)
print(ans2)
```
